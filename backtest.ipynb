```python
# Brent / WTI rolling hedge-ratio model

import numpy as np
import pandas as pd
import yfinance as yf
import statsmodels.api as sm
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller

PRIMARY_SYMBOL = "BZ=F"
HEDGE_SYMBOL = "CL=F"
PRIMARY_LABEL = "Brent"
HEDGE_LABEL = "WTI"

INTERVAL = "1h"
PERIOD = "2y"

BETA_WINDOW = 200
Z_WINDOW = 100
MIN_BETA_ABS = 1e-6

raw = yf.download(
    tickers=[PRIMARY_SYMBOL, HEDGE_SYMBOL],
    interval=INTERVAL,
    period=PERIOD,
    auto_adjust=True,
    progress=False
)

close = raw["Close"][[PRIMARY_SYMBOL, HEDGE_SYMBOL]].copy()
close.columns = ["primary", "hedge"]
close = close.dropna()

df = close.copy()
df["log_primary"] = np.log(df["primary"])
df["log_hedge"] = np.log(df["hedge"])

alphas = []
betas = []
resid_adf_pvalues = []

for i in range(len(df)):
    if i < BETA_WINDOW:
        alphas.append(np.nan)
        betas.append(np.nan)
        resid_adf_pvalues.append(np.nan)
        continue

    window = df.iloc[i - BETA_WINDOW:i]
    y = window["log_primary"]
    x = sm.add_constant(window["log_hedge"])
    model = sm.OLS(y, x).fit()

    alpha = model.params["const"]
    beta = model.params["log_hedge"]
    residuals = y - (alpha + beta * window["log_hedge"])

    try:
        resid_adf_pvalue = adfuller(residuals, regression="c", autolag="AIC")[1]
    except Exception:
        resid_adf_pvalue = np.nan

    alphas.append(alpha)
    betas.append(beta)
    resid_adf_pvalues.append(resid_adf_pvalue)

df["alpha"] = alphas
df["beta_raw"] = betas
df["beta"] = df["beta_raw"].where(df["beta_raw"].abs() >= MIN_BETA_ABS)
df["resid_adf_pvalue"] = resid_adf_pvalues

df["fair_log_primary"] = df["alpha"] + df["beta"] * df["log_hedge"]
df["spread"] = df["log_primary"] - df["fair_log_primary"]
df["spread_mean"] = df["spread"].rolling(Z_WINDOW).mean()
df["spread_std"] = df["spread"].rolling(Z_WINDOW).std()
df["zscore"] = (df["spread"] - df["spread_mean"]) / df["spread_std"]
df = df.replace([np.inf, -np.inf], np.nan).dropna()

latest = df.iloc[-1]
print("===== DATA =====")
print("Start:", df.index.min())
print("End:", df.index.max())
print("Rows:", len(df))
print()
print("===== LATEST MODEL VALUES =====")
print(f"alpha:            {latest['alpha']:.6f}")
print(f"beta:             {latest['beta']:.6f}")
print(f"spread:           {latest['spread']:.6f}")
print(f"z-score:          {latest['zscore']:.2f}")
print(f"resid ADF pvalue: {latest['resid_adf_pvalue']:.4f}")

plt.figure(figsize=(14, 6))
plt.plot(df.index, df["primary"], label=PRIMARY_LABEL)
plt.plot(df.index, df["hedge"], label=HEDGE_LABEL)
plt.title("Brent vs WTI prices")
plt.xlabel("Time")
plt.ylabel("Price")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 4))
plt.plot(df.index, df["beta"], label="Rolling beta")
plt.title(f"Rolling hedge ratio ({BETA_WINDOW} bars)")
plt.xlabel("Time")
plt.ylabel("Beta")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 5))
plt.plot(df.index, df["spread"], label="Spread")
plt.plot(df.index, df["spread_mean"], label="Rolling mean")
plt.title("Log-price spread")
plt.xlabel("Time")
plt.ylabel("Spread")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 5))
plt.plot(df.index, df["zscore"], label="Z-score")
plt.axhline(2.8, linestyle="--", color="tab:red", label="+2.8")
plt.axhline(-2.8, linestyle="--", color="tab:green", label="-2.8")
plt.axhline(0.0, linestyle="-", color="black", label="0")
plt.title("Spread z-score")
plt.xlabel("Time")
plt.ylabel("Z-score")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Brent / WTI backtest with explicit leg-level PnL
# Step 3: choose parameters from config_search.ipynb, run one final backtest, then review Monte Carlo robustness.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----- Trading parameters chosen by the user -----
# ENTRY_Z: enter short spread above +ENTRY_Z, long spread below -ENTRY_Z.
# TP_Z and TP_TOLERANCE: exit when z-score returns near TP_Z within this tolerance band.
# STOP_Z: exit if abs(z-score) reaches this adverse extreme.
# MAX_HOLD_DAYS: time stop in calendar days.
# MAX_RESID_ADF_PVALUE: only trade while rolling residual ADF p-value is at or below this threshold.
ENTRY_Z = 2.8
TP_Z = 0.0
TP_TOLERANCE = 0.15
STOP_Z = 8.0
MAX_HOLD_DAYS = 10
MAX_RESID_ADF_PVALUE = 0.10

# ----- Capital, cost, and sizing -----
INITIAL_CAPITAL = 10_000.0
BASE_PORTFOLIO_FRACTION = 0.10     # Gross exposure as fraction of current equity at entry.
MAX_PORTFOLIO_FRACTION = 0.80      # Absolute cap on gross exposure as fraction of current equity.
FEE_RATE = 0.00015                             # One-way fee fraction of traded notional.
SLIPPAGE_RATE = 0.0                            # One-way slippage fraction of traded notional.

# Position scaling is OFF by default. The model must not assume larger z-score means larger size.
# If enabled, size can increase only according to these explicit user settings.
POSITION_SCALING_ENABLED = True
POSITION_SCALING_STEP_Z = 1.0                  # One size step per this many z-score points beyond ENTRY_Z.
POSITION_SCALING_MODE = "exponential"               # "linear" or "exponential".
POSITION_SCALING_INCREASE = 2.0                # Linear: add 1.0x base per step. Exponential: multiply by this factor per step.

# ----- Monte Carlo settings -----
# Resamples realized trade PnLs to estimate sequence risk. It does not prove future profitability.
MC_SIMULATIONS = 5000
MC_RANDOM_SEED = 42
MC_SAMPLE_CURVES_TO_PLOT = 100

def calculate_target_fraction(z):
    """Return gross exposure fraction. Scaling only happens when POSITION_SCALING_ENABLED is True."""
    abs_z = abs(z)
    if abs_z < ENTRY_Z:
        return 0.0

    if not POSITION_SCALING_ENABLED:
        return min(BASE_PORTFOLIO_FRACTION, MAX_PORTFOLIO_FRACTION)

    steps = int(np.floor((abs_z - ENTRY_Z) / POSITION_SCALING_STEP_Z))
    if POSITION_SCALING_MODE == "linear":
        multiplier = 1.0 + POSITION_SCALING_INCREASE * steps
    elif POSITION_SCALING_MODE == "exponential":
        multiplier = POSITION_SCALING_INCREASE ** steps
    else:
        raise ValueError("POSITION_SCALING_MODE must be 'linear' or 'exponential'.")

    return min(BASE_PORTFOLIO_FRACTION * multiplier, MAX_PORTFOLIO_FRACTION)

def get_leg_targets(gross_notional, beta, primary_price, hedge_price, spread_direction):
    hedge_ratio_abs = abs(beta)
    denom = 1.0 + hedge_ratio_abs
    primary_notional = gross_notional / denom
    hedge_notional = gross_notional - primary_notional

    if spread_direction == 1:
        primary_units = primary_notional / primary_price
        hedge_units = -hedge_notional / hedge_price
    elif spread_direction == -1:
        primary_units = -primary_notional / primary_price
        hedge_units = hedge_notional / hedge_price
    else:
        primary_units = 0.0
        hedge_units = 0.0

    return primary_units, hedge_units

bt = df.copy().sort_index()
bt["primary_price_diff"] = bt["primary"].diff()
bt["hedge_price_diff"] = bt["hedge"].diff()
bt = bt.dropna().copy()

equity = INITIAL_CAPITAL
position = 0
primary_units = 0.0
hedge_units = 0.0
entry_time = None
entry_gross_notional = 0.0
current_trade = None

records = []
trade_log = []

for i in range(1, len(bt)):
    row = bt.iloc[i]
    timestamp = bt.index[i]
    primary_price = row["primary"]
    hedge_price = row["hedge"]
    beta = row["beta"]
    z = row["zscore"]
    resid_adf_pvalue = row["resid_adf_pvalue"]

    action = "HOLD"
    exit_reason = None

    if position != 0:
        pnl = primary_units * row["primary_price_diff"] + hedge_units * row["hedge_price_diff"]
        equity += pnl

        holding_days = (timestamp - entry_time).total_seconds() / (60 * 60 * 24)

        if abs(z - TP_Z) <= TP_TOLERANCE:
            exit_reason = "TP_MEAN_REVERSION"
        elif abs(z) >= STOP_Z:
            exit_reason = "Z_STOP"
        elif holding_days >= MAX_HOLD_DAYS:
            exit_reason = "TIME_STOP"
        elif resid_adf_pvalue > MAX_RESID_ADF_PVALUE:
            exit_reason = "STATIONARITY_FAIL"

        if exit_reason is not None:
            trade_notional = abs(primary_units) * primary_price + abs(hedge_units) * hedge_price
            close_cost = trade_notional * (FEE_RATE + SLIPPAGE_RATE)
            equity -= close_cost

            current_trade["exit_time"] = timestamp
            current_trade["exit_z"] = z
            current_trade["exit_equity"] = equity
            current_trade["exit_reason"] = exit_reason
            current_trade["pnl"] = equity - current_trade["entry_equity"]
            current_trade["pnl_pct"] = equity / current_trade["entry_equity"] - 1.0
            current_trade["exit_gross_notional"] = trade_notional
            trade_log.append(current_trade)

            position = 0
            primary_units = 0.0
            hedge_units = 0.0
            entry_time = None
            entry_gross_notional = 0.0
            current_trade = None
            action = exit_reason

    can_trade = (
        np.isfinite(beta)
        and np.isfinite(z)
        and np.isfinite(resid_adf_pvalue)
        and resid_adf_pvalue <= MAX_RESID_ADF_PVALUE
    )

    if position == 0 and can_trade:
        if z > ENTRY_Z:
            position = -1
        elif z < -ENTRY_Z:
            position = 1

        if position != 0:
            target_fraction = calculate_target_fraction(z)
            target_gross_notional = equity * target_fraction
            primary_units, hedge_units = get_leg_targets(
                target_gross_notional,
                beta,
                primary_price,
                hedge_price,
                position
            )
            entry_cost = target_gross_notional * (FEE_RATE + SLIPPAGE_RATE)
            equity -= entry_cost
            entry_time = timestamp
            entry_gross_notional = target_gross_notional
            current_trade = {
                "entry_time": timestamp,
                "direction": "LONG_SPREAD" if position == 1 else "SHORT_SPREAD",
                "entry_z": z,
                "entry_equity": equity,
                "entry_gross_notional": target_gross_notional,
                "entry_target_fraction": target_fraction,
                "entry_beta": beta,
                "entry_resid_adf_pvalue": resid_adf_pvalue
            }
            action = current_trade["direction"]

    elif position != 0 and can_trade and POSITION_SCALING_ENABLED:
        target_fraction = calculate_target_fraction(z)
        target_gross_notional = equity * target_fraction

        current_gross_notional = abs(primary_units) * primary_price + abs(hedge_units) * hedge_price
        if target_gross_notional > current_gross_notional:
            target_primary_units, target_hedge_units = get_leg_targets(
                target_gross_notional,
                beta,
                primary_price,
                hedge_price,
                position
            )
            delta_primary_units = target_primary_units - primary_units
            delta_hedge_units = target_hedge_units - hedge_units
            traded_notional = abs(delta_primary_units) * primary_price + abs(delta_hedge_units) * hedge_price

            primary_units = target_primary_units
            hedge_units = target_hedge_units
            equity -= traded_notional * (FEE_RATE + SLIPPAGE_RATE)
            action = f"INCREASE_TO_{target_fraction * 100:.2f}_PCT"

    gross_notional = abs(primary_units) * primary_price + abs(hedge_units) * hedge_price
    net_notional = primary_units * primary_price + hedge_units * hedge_price

    records.append({
        "time": timestamp,
        "equity": equity,
        "position": position,
        "gross_notional": gross_notional,
        "net_notional": net_notional,
        "primary_units": primary_units,
        "hedge_units": hedge_units,
        "zscore": z,
        "spread": row["spread"],
        "resid_adf_pvalue": resid_adf_pvalue,
        "action": action
    })

result = pd.DataFrame(records).set_index("time")
trades = pd.DataFrame(trade_log)

bars_per_year = 252 * 24 if str(INTERVAL).endswith("h") else 252
returns = result["equity"].pct_change().dropna()
sharpe = np.nan
omega = np.nan

if len(returns) > 1 and returns.std() > 0:
    sharpe = (returns.mean() / returns.std()) * np.sqrt(bars_per_year)

threshold = 0.0
gains = (returns - threshold).clip(lower=0).sum()
losses = (threshold - returns).clip(lower=0).sum()
if losses > 0:
    omega = gains / losses

result["equity_peak"] = result["equity"].cummax()
result["drawdown"] = result["equity"] / result["equity_peak"] - 1.0

print("===== BACKTEST PARAMETERS =====")
print(f"ENTRY_Z={ENTRY_Z}, STOP_Z={STOP_Z}, TP_Z={TP_Z}, TP_TOLERANCE={TP_TOLERANCE}, MAX_HOLD_DAYS={MAX_HOLD_DAYS}")
print(f"MAX_RESID_ADF_PVALUE={MAX_RESID_ADF_PVALUE}")
print(f"BASE_PORTFOLIO_FRACTION={BASE_PORTFOLIO_FRACTION:.4f}, MAX_PORTFOLIO_FRACTION={MAX_PORTFOLIO_FRACTION:.4f}")
print(f"POSITION_SCALING_ENABLED={POSITION_SCALING_ENABLED}")
if POSITION_SCALING_ENABLED:
    print(f"Scaling mode={POSITION_SCALING_MODE}, step z={POSITION_SCALING_STEP_Z}, increase={POSITION_SCALING_INCREASE}")
else:
    print("Sizing uses the fixed base fraction. Larger z-score does not automatically increase size.")

print("===== BACKTEST SUMMARY =====")
print(f"Start Equity: {INITIAL_CAPITAL:.2f}")
print(f"End Equity:   {result['equity'].iloc[-1]:.2f}")
print(f"Return:       {(result['equity'].iloc[-1] / INITIAL_CAPITAL - 1) * 100:.2f}%")
print(f"Trades:       {len(trades)}")
print(f"Sharpe:       {sharpe:.2f}")
print(f"Omega:        {omega:.2f}")
print(f"Max Drawdown: {result['drawdown'].min() * 100:.2f}%")

if len(trades) > 0:
    print(f"Winrate:      {(trades['pnl'] > 0).mean() * 100:.2f}%")
    print(f"Avg PnL:      {trades['pnl'].mean():.2f}")
    print(f"Best Trade:   {trades['pnl'].max():.2f}")
    print(f"Worst Trade:  {trades['pnl'].min():.2f}")
    display(trades.tail(20))
else:
    print("No closed trades. Monte Carlo simulation is skipped.")

print("===== OUTPUT GUIDE =====")
print("- Equity is account value after mark-to-market PnL and trading costs.")
print("- Gross notional is abs(long leg) + abs(short leg), not net exposure.")
print("- Max drawdown is the worst percentage fall from a prior equity peak.")
print("- Monte Carlo below resamples realized trade PnL to show sequence risk, not a prediction.")

plt.figure(figsize=(14, 6))
plt.plot(result.index, result["equity"], label="Equity")
plt.title("Brent / WTI equity curve")
plt.xlabel("Time")
plt.ylabel("Equity USD")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 4))
plt.plot(result.index, result["drawdown"] * 100, label="Drawdown %")
plt.title("Drawdown")
plt.xlabel("Time")
plt.ylabel("Drawdown %")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 6))
plt.plot(result.index, result["zscore"], label="Z-score")
plt.axhline(ENTRY_Z, linestyle="--", label=f"+{ENTRY_Z} entry")
plt.axhline(-ENTRY_Z, linestyle="--", label=f"-{ENTRY_Z} entry")
plt.axhline(0.0, linestyle="-", label="0")
plt.scatter(result.index, result["zscore"], s=8, alpha=0.25)
plt.title("Z-score and trade timing")
plt.xlabel("Time")
plt.ylabel("Z-score")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 4))
plt.plot(result.index, result["gross_notional"], label="Gross notional")
plt.title("Gross exposure")
plt.xlabel("Time")
plt.ylabel("USD")
plt.grid(True)
plt.legend()
plt.show()

if len(trades) > 0:
    rng = np.random.default_rng(MC_RANDOM_SEED)
    trade_pnls = trades["pnl"].to_numpy(dtype=float)
    sampled_pnls = rng.choice(trade_pnls, size=(MC_SIMULATIONS, len(trade_pnls)), replace=True)
    mc_equity_paths = INITIAL_CAPITAL + np.cumsum(sampled_pnls, axis=1)
    mc_ending_equity = mc_equity_paths[:, -1]
    mc_returns_pct = (mc_ending_equity / INITIAL_CAPITAL - 1.0) * 100.0
    mc_running_peaks = np.maximum.accumulate(mc_equity_paths, axis=1)
    mc_drawdowns = mc_equity_paths / mc_running_peaks - 1.0
    mc_max_drawdowns_pct = mc_drawdowns.min(axis=1) * 100.0

    print("===== MONTE CARLO TRADE-PNL RESAMPLING =====")
    print(f"Simulations: {MC_SIMULATIONS}, trades per simulation: {len(trade_pnls)}")
    print(f"Ending equity 5/50/95 pct: {np.percentile(mc_ending_equity, 5):.2f} / {np.percentile(mc_ending_equity, 50):.2f} / {np.percentile(mc_ending_equity, 95):.2f}")
    print(f"Return 5/50/95 pct:        {np.percentile(mc_returns_pct, 5):.2f}% / {np.percentile(mc_returns_pct, 50):.2f}% / {np.percentile(mc_returns_pct, 95):.2f}%")
    print(f"Max DD 5/50/95 pct:        {np.percentile(mc_max_drawdowns_pct, 5):.2f}% / {np.percentile(mc_max_drawdowns_pct, 50):.2f}% / {np.percentile(mc_max_drawdowns_pct, 95):.2f}%")
    print(f"Probability ending below start: {(mc_ending_equity < INITIAL_CAPITAL).mean() * 100:.2f}%")

    plt.figure(figsize=(14, 6))
    sample_count = min(MC_SAMPLE_CURVES_TO_PLOT, MC_SIMULATIONS)
    for path in mc_equity_paths[:sample_count]:
        plt.plot(range(1, len(path) + 1), path, color="0.70", alpha=0.25, linewidth=0.8)
    plt.plot(range(1, len(trade_pnls) + 1), INITIAL_CAPITAL + np.cumsum(trade_pnls), color="tab:blue", linewidth=2.5, label="Actual trade order")
    plt.title("Monte Carlo equity paths from resampled trade PnL")
    plt.xlabel("Trade number")
    plt.ylabel("Equity USD")
    plt.grid(True)
    plt.legend()
    plt.show()

    plt.figure(figsize=(14, 4))
    plt.hist(mc_returns_pct, bins=50, alpha=0.8)
    plt.axvline((result["equity"].iloc[-1] / INITIAL_CAPITAL - 1.0) * 100.0, color="tab:blue", linestyle="--", label="Actual return")
    plt.title("Monte Carlo ending return distribution")
    plt.xlabel("Ending return %")
    plt.ylabel("Simulation count")
    plt.grid(True)
    plt.legend()
    plt.show()
```
