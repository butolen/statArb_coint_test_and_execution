```python
# Brent / WTI execution notebook using the same logic as the backtest

import numpy as np
import pandas as pd
import yfinance as yf
import statsmodels.api as sm
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller

PRIMARY_SYMBOL = "BZ=F"
HEDGE_SYMBOL = "CL=F"
PRIMARY_LABEL = "Brent"
HEDGE_LABEL = "WTI"

INTERVAL = "1h"
PERIOD = "729d"

BETA_WINDOW = 200
Z_WINDOW = 100
ENTRY_Z = 2.8
TP_Z = 0.0
TP_TOLERANCE = 0.15
STOP_Z = 8.0
MAX_HOLD_DAYS = 10
MAX_RESID_ADF_PVALUE = 0.1

# ----- Live account and sizing inputs -----
# PORTFOLIO_EQUITY_USD: account equity used to convert target fraction into gross notional.
# BASE_PORTFOLIO_FRACTION: default gross exposure fraction for a fresh signal.
# MAX_PORTFOLIO_FRACTION: absolute exposure cap even if explicit scaling is enabled.
PORTFOLIO_EQUITY_USD = 10_000.0
BASE_PORTFOLIO_FRACTION = 0.10
MAX_PORTFOLIO_FRACTION = 0.80

# Position scaling is OFF by default. Do not assume a larger z-score means a larger live order.
# If enabled, the increase is controlled by the step size, mode, and increase value below.
POSITION_SCALING_ENABLED = False
POSITION_SCALING_STEP_Z = 1.0
POSITION_SCALING_MODE = "linear"      # "linear" or "exponential".
POSITION_SCALING_INCREASE = 1.0       # Linear: add 1.0x base per step. Exponential: multiply by this factor per step.

FEE_RATE = 0.00015
SLIPPAGE_RATE = 0.0
LOOKBACK_PLOT = 300
MIN_BETA_ABS = 1e-6

def calculate_target_fraction(z):
    """Return gross exposure fraction. Scaling only happens when explicitly enabled above."""
    abs_z = abs(z)
    if abs_z < ENTRY_Z:
        return 0.0

    if not POSITION_SCALING_ENABLED:
        return min(BASE_PORTFOLIO_FRACTION, MAX_PORTFOLIO_FRACTION)

    steps = int(np.floor((abs_z - ENTRY_Z) / POSITION_SCALING_STEP_Z))
    if POSITION_SCALING_MODE == "linear":
        multiplier = 1.0 + POSITION_SCALING_INCREASE * steps
    elif POSITION_SCALING_MODE == "exponential":
        multiplier = POSITION_SCALING_INCREASE ** steps
    else:
        raise ValueError("POSITION_SCALING_MODE must be 'linear' or 'exponential'.")

    return min(BASE_PORTFOLIO_FRACTION * multiplier, MAX_PORTFOLIO_FRACTION)

def get_signal(z, resid_adf_pvalue):
    if not np.isfinite(resid_adf_pvalue) or resid_adf_pvalue > MAX_RESID_ADF_PVALUE:
        return "NO_TRADE_STATIONARITY_FAIL"
    if z > ENTRY_Z:
        return "SHORT_SPREAD"
    if z < -ENTRY_Z:
        return "LONG_SPREAD"
    if abs(z - TP_Z) <= TP_TOLERANCE:
        return "TAKE_PROFIT_ZONE"
    return "NO_TRADE"

def get_leg_plan(gross_notional, beta, primary_price, hedge_price, spread_direction):
    hedge_ratio_abs = abs(beta)
    primary_notional = gross_notional / (1.0 + hedge_ratio_abs)
    hedge_notional = gross_notional - primary_notional
    primary_size = primary_notional / primary_price
    hedge_size = hedge_notional / hedge_price

    if spread_direction == "LONG_SPREAD":
        primary_side = "LONG"
        hedge_side = "SHORT"
    else:
        primary_side = "SHORT"
        hedge_side = "LONG"

    return {
        "primary_side": primary_side,
        "hedge_side": hedge_side,
        "primary_notional": primary_notional,
        "hedge_notional": hedge_notional,
        "primary_size": primary_size,
        "hedge_size": hedge_size
    }

raw = yf.download(
    tickers=[PRIMARY_SYMBOL, HEDGE_SYMBOL],
    interval=INTERVAL,
    period=PERIOD,
    auto_adjust=True,
    progress=False
)

close = raw["Close"][[PRIMARY_SYMBOL, HEDGE_SYMBOL]].copy()
close.columns = ["primary", "hedge"]
close = close.dropna()

df_exec = close.copy()
df_exec["log_primary"] = np.log(df_exec["primary"])
df_exec["log_hedge"] = np.log(df_exec["hedge"])

alphas = []
betas = []
resid_adf_pvalues = []

for i in range(len(df_exec)):
    if i < BETA_WINDOW:
        alphas.append(np.nan)
        betas.append(np.nan)
        resid_adf_pvalues.append(np.nan)
        continue

    window = df_exec.iloc[i - BETA_WINDOW:i]
    y = window["log_primary"]
    x = sm.add_constant(window["log_hedge"])
    model = sm.OLS(y, x).fit()
    alpha = model.params["const"]
    beta = model.params["log_hedge"]
    residuals = y - (alpha + beta * window["log_hedge"])

    try:
        resid_adf_pvalue = adfuller(residuals, regression="c", autolag="AIC")[1]
    except Exception:
        resid_adf_pvalue = np.nan

    alphas.append(alpha)
    betas.append(beta)
    resid_adf_pvalues.append(resid_adf_pvalue)

df_exec["alpha"] = alphas
df_exec["beta_raw"] = betas
df_exec["beta"] = df_exec["beta_raw"].where(df_exec["beta_raw"].abs() >= MIN_BETA_ABS)
df_exec["resid_adf_pvalue"] = resid_adf_pvalues
df_exec["fair_log_primary"] = df_exec["alpha"] + df_exec["beta"] * df_exec["log_hedge"]
df_exec["spread"] = df_exec["log_primary"] - df_exec["fair_log_primary"]
df_exec["spread_mean"] = df_exec["spread"].rolling(Z_WINDOW).mean()
df_exec["spread_std"] = df_exec["spread"].rolling(Z_WINDOW).std()
df_exec["zscore"] = (df_exec["spread"] - df_exec["spread_mean"]) / df_exec["spread_std"]
df_exec = df_exec.replace([np.inf, -np.inf], np.nan).dropna()

latest = df_exec.iloc[-1]
current_time = df_exec.index[-1]
primary_price = latest["primary"]
hedge_price = latest["hedge"]
alpha = latest["alpha"]
beta = latest["beta"]
spread = latest["spread"]
zscore = latest["zscore"]
resid_adf_pvalue = latest["resid_adf_pvalue"]

signal = get_signal(zscore, resid_adf_pvalue)
target_fraction = calculate_target_fraction(zscore)
gross_notional = PORTFOLIO_EQUITY_USD * target_fraction

print("===== CURRENT EXECUTION SIGNAL =====")
print(f"Time:               {current_time}")
print(f"Primary:            {PRIMARY_SYMBOL} ({PRIMARY_LABEL})")
print(f"Hedge:              {HEDGE_SYMBOL} ({HEDGE_LABEL})")
print(f"Primary Price:      {primary_price:.4f}")
print(f"Hedge Price:        {hedge_price:.4f}")
print(f"Alpha:              {alpha:.6f}")
print(f"Beta:               {beta:.6f}")
print(f"Spread:             {spread:.6f}")
print(f"Z-score:            {zscore:.2f}")
print(f"Residual ADF pvalue:{resid_adf_pvalue:.4f}")
print(f"Signal:             {signal}")
print(f"Target Fraction:    {target_fraction * 100:.2f}%")
print(f"Gross Notional:     ${gross_notional:,.2f}")
print(f"Position Scaling:   {POSITION_SCALING_ENABLED}")
if POSITION_SCALING_ENABLED:
    print(f"Scaling Rule:       {POSITION_SCALING_MODE}, step z {POSITION_SCALING_STEP_Z}, increase {POSITION_SCALING_INCREASE}")
else:
    print("Scaling Rule:       fixed base fraction; z-score does not automatically compound size")

if signal in ["LONG_SPREAD", "SHORT_SPREAD"]:
    plan = get_leg_plan(gross_notional, beta, primary_price, hedge_price, signal)
    print("===== EXECUTION PLAN =====")
    print(f"{plan['primary_side']} {PRIMARY_LABEL}: ${plan['primary_notional']:,.2f} | size {plan['primary_size']:.6f}")
    print(f"{plan['hedge_side']} {HEDGE_LABEL}: ${plan['hedge_notional']:,.2f} | size {plan['hedge_size']:.6f}")
elif signal == "TAKE_PROFIT_ZONE":
    print("===== EXECUTION PLAN =====")
    print("Spread is back near the mean. Close existing spread exposure.")
else:
    print("===== EXECUTION PLAN =====")
    print("No new trade.")

plot_df = df_exec.tail(LOOKBACK_PLOT)

plt.figure(figsize=(14, 6))
plt.plot(plot_df.index, plot_df["zscore"], label="Z-score")
plt.axhline(ENTRY_Z, linestyle="--", label=f"+{ENTRY_Z} entry")
plt.axhline(-ENTRY_Z, linestyle="--", label=f"-{ENTRY_Z} entry")
plt.axhline(0.0, linestyle="-", label="0")
plt.scatter(current_time, zscore, s=120, label=f"Current z {zscore:.2f}")
plt.title("Brent / WTI z-score")
plt.xlabel("Time")
plt.ylabel("Z-score")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 5))
plt.plot(plot_df.index, plot_df["spread"], label="Spread")
plt.plot(plot_df.index, plot_df["spread_mean"], label="Rolling mean")
plt.scatter(current_time, spread, s=120, label="Current spread")
plt.title("Spread")
plt.xlabel("Time")
plt.ylabel("Spread")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 4))
plt.plot(plot_df.index, plot_df["beta"], label="Rolling beta")
plt.scatter(current_time, beta, s=120, label=f"Current beta {beta:.4f}")
plt.title("Rolling hedge ratio")
plt.xlabel("Time")
plt.ylabel("Beta")
plt.grid(True)
plt.legend()
plt.show()
```
