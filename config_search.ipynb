```python
# Step 2: configuration search for one user-selected timeframe
# Run coint_test.ipynb first. Only search configs after the pair has acceptable cointegration diagnostics.

import itertools
import math
import numpy as np
import pandas as pd
import yfinance as yf
import statsmodels.api as sm
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller

# ----- Pair selection -----
# PRIMARY_SYMBOL is the instrument modeled as "fair value = alpha + beta * hedge".
# HEDGE_SYMBOL is the instrument used to hedge PRIMARY_SYMBOL.
PRIMARY_SYMBOL = "BZ=F"
HEDGE_SYMBOL = "CL=F"
PRIMARY_LABEL = "Brent"
HEDGE_LABEL = "WTI"

# ----- Data selection -----
# Search exactly one timeframe. Do not optimize across intervals in this notebook.
# Examples: "1h", "2h", "4h", "1d". Yahoo period must support the interval.
USER_INTERVAL = "1h"
USER_PERIOD = "2y"

# ----- Model search grid -----
# BETA_WINDOWS: bars used for rolling OLS hedge ratio. Higher = smoother/slower, lower = faster/noisier.
# Z_WINDOWS: bars used for spread mean/std z-score. Higher = smoother, lower = more signals.
BETA_WINDOWS = [120, 200, 300]
Z_WINDOWS = [60, 100, 150]

# ----- Trading rule search grid -----
# ENTRY_Z_OPTIONS: enter when abs(z-score) is above this value.
# STOP_Z_OPTIONS: force exit if abs(z-score) keeps moving against mean reversion.
# MAX_HOLD_BARS_OPTIONS: time stop measured in bars of USER_INTERVAL.
# MAX_RESID_ADF_PVALUE_OPTIONS: max rolling residual ADF p-value allowed for entries/holds.
ENTRY_Z_OPTIONS = [2.0, 2.4, 2.8]
STOP_Z_OPTIONS = [4.5, 5.0, 5.5]
MAX_HOLD_BARS_OPTIONS = [24, 48, 96,10,14]
MAX_RESID_ADF_PVALUE_OPTIONS = [0.05, 0.10]

# ----- Exit rule constants -----
# TP_Z is the mean-reversion target. 0.0 means exit near the rolling spread mean.
# TP_TOLERANCE allows exits inside +/- this z-score band around TP_Z.
TP_Z = 0.0
TP_TOLERANCE = 0.15

# ----- Capital, cost, and sizing -----
INITIAL_CAPITAL = 10_000.0
BASE_PORTFOLIO_FRACTION = 0.1     # Default gross exposure as fraction of equity at entry.
MAX_PORTFOLIO_FRACTION = 0.8       # Hard cap on gross exposure as fraction of equity.
FEE_RATE = 0.00015                  # One-way commission fraction of traded notional.
SLIPPAGE_RATE = 0.0                 # One-way slippage fraction of traded notional.

# Position scaling is OFF by default. If False, every qualifying entry uses BASE_PORTFOLIO_FRACTION.
# If True, size can increase as abs(z-score) moves farther beyond ENTRY_Z.
POSITION_SCALING_ENABLED = False
POSITION_SCALING_STEP_Z = 1.0       # One size step for every this many z-score points beyond ENTRY_Z.
POSITION_SCALING_MODE = "exponential"    # "linear" or "exponential".
POSITION_SCALING_INCREASE = 2.0     # Linear: add 1.0x base per step. Exponential: multiply by this factor per step.

# ----- Safety filters and display -----
MIN_BETA_ABS = 1e-6                 # Avoid division/exposure issues when beta is effectively zero.
MIN_TRADES = 20                     # Minimum trades required for a config to pass the robustness filter.
TOP_N = 25
PLOT_ALL_EQUITY_CURVES = True       # Plots every tested config in gray; top configs are highlighted.
MAX_EQUITY_CURVES_TO_LABEL = 10

def calculate_target_fraction(z, entry_z):
    """Return gross exposure fraction. Scaling only happens when explicitly enabled above."""
    abs_z = abs(z)
    if abs_z < entry_z:
        return 0.0

    if not POSITION_SCALING_ENABLED:
        return min(BASE_PORTFOLIO_FRACTION, MAX_PORTFOLIO_FRACTION)

    steps = int(np.floor((abs_z - entry_z) / POSITION_SCALING_STEP_Z))
    if POSITION_SCALING_MODE == "linear":
        multiplier = 1.0 + POSITION_SCALING_INCREASE * steps
    elif POSITION_SCALING_MODE == "exponential":
        multiplier = POSITION_SCALING_INCREASE ** steps
    else:
        raise ValueError("POSITION_SCALING_MODE must be 'linear' or 'exponential'.")

    return min(BASE_PORTFOLIO_FRACTION * multiplier, MAX_PORTFOLIO_FRACTION)

def safe_adf(series):
    series = pd.Series(series).dropna()
    if len(series) < 25:
        return np.nan
    try:
        return adfuller(series, regression="c", autolag="AIC")[1]
    except Exception:
        return np.nan

def get_leg_targets(gross_notional, beta, primary_price, hedge_price, spread_direction):
    hedge_ratio_abs = abs(beta)
    denom = 1.0 + hedge_ratio_abs
    primary_notional = gross_notional / denom
    hedge_notional = gross_notional - primary_notional

    if spread_direction == 1:
        primary_units = primary_notional / primary_price
        hedge_units = -hedge_notional / hedge_price
    else:
        primary_units = -primary_notional / primary_price
        hedge_units = hedge_notional / hedge_price

    return primary_units, hedge_units

def build_dataset(interval, period, beta_window, z_window):
    raw = yf.download(
        tickers=[PRIMARY_SYMBOL, HEDGE_SYMBOL],
        interval=interval,
        period=period,
        auto_adjust=True,
        progress=False
    )

    close = raw["Close"][[PRIMARY_SYMBOL, HEDGE_SYMBOL]].copy()
    close.columns = ["primary", "hedge"]
    close = close.dropna()

    if len(close) < max(beta_window, z_window) + 50:
        return None

    df = close.copy()
    df["log_primary"] = np.log(df["primary"])
    df["log_hedge"] = np.log(df["hedge"])

    alphas = []
    betas = []
    resid_adf_pvalues = []

    for i in range(len(df)):
        if i < beta_window:
            alphas.append(np.nan)
            betas.append(np.nan)
            resid_adf_pvalues.append(np.nan)
            continue

        window = df.iloc[i - beta_window:i]
        y = window["log_primary"]
        x = sm.add_constant(window["log_hedge"])
        model = sm.OLS(y, x).fit()
        alpha = model.params["const"]
        beta = model.params["log_hedge"]
        residuals = y - (alpha + beta * window["log_hedge"])

        alphas.append(alpha)
        betas.append(beta)
        resid_adf_pvalues.append(safe_adf(residuals))

    df["alpha"] = alphas
    df["beta_raw"] = betas
    df["beta"] = df["beta_raw"].where(df["beta_raw"].abs() >= MIN_BETA_ABS)
    df["resid_adf_pvalue"] = resid_adf_pvalues
    df["fair_log_primary"] = df["alpha"] + df["beta"] * df["log_hedge"]
    df["spread"] = df["log_primary"] - df["fair_log_primary"]
    df["spread_mean"] = df["spread"].rolling(z_window).mean()
    df["spread_std"] = df["spread"].rolling(z_window).std()
    df["zscore"] = (df["spread"] - df["spread_mean"]) / df["spread_std"]
    df["primary_price_diff"] = df["primary"].diff()
    df["hedge_price_diff"] = df["hedge"].diff()
    df = df.replace([np.inf, -np.inf], np.nan).dropna().copy()

    return df if len(df) > 100 else None

def backtest(df, interval, entry_z, stop_z, max_hold_bars, max_resid_adf_pvalue):
    equity = INITIAL_CAPITAL
    position = 0
    primary_units = 0.0
    hedge_units = 0.0
    entry_idx = None
    trade_pnls = []
    equity_points = []
    equity_index = []
    actions = 0
    stationarity_pass_bars = 0
    valid_signal_bars = 0
    entry_equity = None

    for i in range(1, len(df)):
        row = df.iloc[i]
        timestamp = df.index[i]
        primary_price = row["primary"]
        hedge_price = row["hedge"]
        beta = row["beta"]
        z = row["zscore"]
        resid_adf_pvalue = row["resid_adf_pvalue"]

        can_trade = np.isfinite(beta) and np.isfinite(z) and np.isfinite(resid_adf_pvalue)
        if can_trade:
            valid_signal_bars += 1
            if resid_adf_pvalue <= max_resid_adf_pvalue:
                stationarity_pass_bars += 1

        if position != 0:
            pnl = primary_units * row["primary_price_diff"] + hedge_units * row["hedge_price_diff"]
            equity += pnl

            holding_bars = i - entry_idx
            exit_reason = None
            if abs(z - TP_Z) <= TP_TOLERANCE:
                exit_reason = "TP"
            elif abs(z) >= stop_z:
                exit_reason = "STOP"
            elif holding_bars >= max_hold_bars:
                exit_reason = "TIME"
            elif resid_adf_pvalue > max_resid_adf_pvalue:
                exit_reason = "ADF_FAIL"

            if exit_reason is not None:
                gross_notional = abs(primary_units) * primary_price + abs(hedge_units) * hedge_price
                equity -= gross_notional * (FEE_RATE + SLIPPAGE_RATE)
                trade_pnls.append(equity - entry_equity)
                position = 0
                primary_units = 0.0
                hedge_units = 0.0
                entry_idx = None
                entry_equity = None
                actions += 1

        if position == 0 and can_trade and resid_adf_pvalue <= max_resid_adf_pvalue:
            if z > entry_z:
                position = -1
            elif z < -entry_z:
                position = 1

            if position != 0:
                target_fraction = calculate_target_fraction(z, entry_z)
                gross_notional = equity * target_fraction
                primary_units, hedge_units = get_leg_targets(gross_notional, beta, primary_price, hedge_price, position)
                equity -= gross_notional * (FEE_RATE + SLIPPAGE_RATE)
                entry_idx = i
                entry_equity = equity
                actions += 1

        equity_points.append(equity)
        equity_index.append(timestamp)

    if not equity_points:
        return None

    equity_curve = pd.Series(equity_points, index=pd.Index(equity_index, name="time"), name="equity")
    drawdown = equity_curve / equity_curve.cummax() - 1.0
    trade_pnls = pd.Series(trade_pnls, dtype=float)

    returns = equity_curve.pct_change().dropna()
    sharpe = np.nan
    if returns.std() > 0:
        bars_per_year = 252 * 24 if interval.endswith("h") else 252
        sharpe = (returns.mean() / returns.std()) * np.sqrt(bars_per_year)

    winrate = (trade_pnls > 0).mean() if len(trade_pnls) else np.nan
    gross_profit = trade_pnls[trade_pnls > 0].sum()
    gross_loss = -trade_pnls[trade_pnls < 0].sum()
    profit_factor = gross_profit / gross_loss if gross_loss > 0 else np.nan

    return {
        "interval": interval,
        "rows": len(df),
        "entry_z": entry_z,
        "stop_z": stop_z,
        "max_hold_bars": max_hold_bars,
        "max_resid_adf_pvalue": max_resid_adf_pvalue,
        "position_scaling_enabled": POSITION_SCALING_ENABLED,
        "position_scaling_mode": POSITION_SCALING_MODE if POSITION_SCALING_ENABLED else "off",
        "trades": int(len(trade_pnls)),
        "return_pct": (equity_curve.iloc[-1] / INITIAL_CAPITAL - 1.0) * 100.0,
        "end_equity": equity_curve.iloc[-1],
        "avg_trade_pnl": trade_pnls.mean() if len(trade_pnls) else np.nan,
        "median_trade_pnl": trade_pnls.median() if len(trade_pnls) else np.nan,
        "winrate_pct": winrate * 100.0 if pd.notna(winrate) else np.nan,
        "profit_factor": profit_factor,
        "max_drawdown_pct": drawdown.min() * 100.0,
        "sharpe": sharpe,
        "stationarity_pass_ratio_pct": (stationarity_pass_bars / valid_signal_bars * 100.0) if valid_signal_bars else np.nan,
        "actions": actions,
        "equity_curve": equity_curve
    }

print(f"Loading one search timeframe: interval={USER_INTERVAL}, period={USER_PERIOD}")
datasets = {}
for beta_window, z_window in itertools.product(BETA_WINDOWS, Z_WINDOWS):
    key = (USER_INTERVAL, USER_PERIOD, beta_window, z_window)
    print(f"Loading {key} ...")
    datasets[key] = build_dataset(USER_INTERVAL, USER_PERIOD, beta_window, z_window)
```

```python
# Run the single-timeframe search, rank configurations, and plot equity curves

results = []
equity_curves = {}

for (interval, period, beta_window, z_window), df in datasets.items():
    if df is None:
        continue

    for entry_z, stop_z, max_hold_bars, max_resid_adf_pvalue in itertools.product(
        ENTRY_Z_OPTIONS,
        STOP_Z_OPTIONS,
        MAX_HOLD_BARS_OPTIONS,
        MAX_RESID_ADF_PVALUE_OPTIONS
    ):
        res = backtest(
            df=df,
            interval=interval,
            entry_z=entry_z,
            stop_z=stop_z,
            max_hold_bars=max_hold_bars,
            max_resid_adf_pvalue=max_resid_adf_pvalue
        )
        if res is None:
            continue

        curve_key = (
            beta_window,
            z_window,
            entry_z,
            stop_z,
            max_hold_bars,
            max_resid_adf_pvalue
        )
        equity_curves[curve_key] = res.pop("equity_curve")
        res["curve_key"] = curve_key
        res["period"] = period
        res["beta_window"] = beta_window
        res["z_window"] = z_window
        results.append(res)

results_df = pd.DataFrame(results)
if results_df.empty:
    raise ValueError("No valid search results produced.")

filtered = results_df[
    (results_df["trades"] >= MIN_TRADES)
    & (results_df["return_pct"] > 0)
    & (results_df["max_drawdown_pct"] > -10)
    & (results_df["stationarity_pass_ratio_pct"] >= 25)
].copy()

if filtered.empty:
    print("No configs passed the stability filter. Showing all results instead.")
    filtered = results_df.copy()

# Rank by robustness, not only return. Positive drawdown values do not exist; less negative is better.
filtered["rank_score"] = (
    filtered["return_pct"].fillna(-999) * 0.30
    + filtered["sharpe"].fillna(-999) * 0.35
    + filtered["profit_factor"].fillna(0) * 10.0 * 0.20
    + filtered["stationarity_pass_ratio_pct"].fillna(0) * 0.10
    + filtered["trades"].fillna(0) * 0.05
    + filtered["max_drawdown_pct"].fillna(-999) * 0.10
)

top_configs = filtered.sort_values(["rank_score", "return_pct", "sharpe"], ascending=False).head(TOP_N)

print("===== SEARCH SCOPE =====")
print(f"Pair: {PRIMARY_LABEL} ({PRIMARY_SYMBOL}) vs {HEDGE_LABEL} ({HEDGE_SYMBOL})")
print(f"Timeframe searched: {USER_INTERVAL}, period: {USER_PERIOD}")
print(f"Configs tested: {len(results_df)}")
print(f"Position scaling enabled: {POSITION_SCALING_ENABLED}")
if POSITION_SCALING_ENABLED:
    print(f"Scaling mode: {POSITION_SCALING_MODE}, step z: {POSITION_SCALING_STEP_Z}, increase: {POSITION_SCALING_INCREASE}")
else:
    print("Sizing: fixed BASE_PORTFOLIO_FRACTION per entry; no automatic compounding when z-score is larger.")

print("===== TOP CONFIGS =====")
display(top_configs[[
    "interval", "period", "beta_window", "z_window", "entry_z", "stop_z",
    "max_hold_bars", "max_resid_adf_pvalue", "position_scaling_enabled",
    "trades", "return_pct", "winrate_pct", "avg_trade_pnl", "profit_factor",
    "sharpe", "max_drawdown_pct", "stationarity_pass_ratio_pct", "rank_score"
]])

if PLOT_ALL_EQUITY_CURVES:
    plt.figure(figsize=(14, 7))
    for curve_key, curve in equity_curves.items():
        plt.plot(curve.index, curve.values, color="0.75", linewidth=0.8, alpha=0.25)

    for rank, (_, row) in enumerate(top_configs.head(MAX_EQUITY_CURVES_TO_LABEL).iterrows(), start=1):
        curve = equity_curves[row["curve_key"]]
        label = (
            f"#{rank} beta={row['beta_window']} zwin={row['z_window']} "
            f"entry={row['entry_z']} stop={row['stop_z']} hold={row['max_hold_bars']}"
        )
        plt.plot(curve.index, curve.values, linewidth=2.0, label=label)

    plt.title(f"Equity curves for all tested configs ({USER_INTERVAL})")
    plt.xlabel("Time")
    plt.ylabel("Equity USD")
    plt.grid(True)
    plt.legend(fontsize=8)
    plt.show()
```

```python
# Output guide for the configuration search

print("Interpretation guide:")
print("- return_pct is total backtest return for the config on the one selected timeframe.")
print("- sharpe is annualized from bar returns; compare only within the same timeframe.")
print("- profit_factor is gross winning trade PnL divided by gross losing trade PnL. Higher is better, but unstable with few trades.")
print("- max_drawdown_pct is the worst equity decline from a prior peak. Closer to 0 is better.")
print("- stationarity_pass_ratio_pct is how often the rolling residual ADF filter accepted the spread.")
print("- The equity-curve plot is for overfitting checks: prefer configs that behave similarly to neighbors, not one isolated spike.")
print("- If POSITION_SCALING_ENABLED is False, larger z-score entries do not automatically get larger position size.")
```
