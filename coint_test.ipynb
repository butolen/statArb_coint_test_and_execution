```python
# Step 1: cointegration test notebook
# Purpose: decide whether the pair is statistically suitable before running config_search.ipynb.

import numpy as np
import pandas as pd
import yfinance as yf
import statsmodels.api as sm
import matplotlib.pyplot as plt

from statsmodels.tsa.stattools import adfuller, coint
from statsmodels.tsa.vector_ar.vecm import coint_johansen

# ----- Pair selection -----
# PRIMARY_SYMBOL is modeled as: log(primary) = alpha + beta * log(hedge) + residual.
# HEDGE_SYMBOL is the hedge leg used to explain primary.
# Labels only affect plot and print names.
PRIMARY_SYMBOL = "BZ=F"
HEDGE_SYMBOL = "CL=F"
PRIMARY_LABEL = "Primary"
HEDGE_LABEL = "Hedge"

# ----- Data selection -----
# INTERVAL is the bar size downloaded from Yahoo, e.g. "1h", "2h", "4h", "1d".
# PERIOD is the amount of history, e.g. "2y" for intraday or "10y" for daily.
INTERVAL = "1h"
PERIOD = "2y"

# ----- Rolling diagnostics -----
# ROLLING_WINDOW: bars per rolling OLS/ADF window. Should be long enough to estimate beta reliably.
# ROLLING_STEP: bars between rolling test windows. Lower = more detail and slower runtime.
# MAX_RESID_ADF_PVALUE: residual stationarity threshold. 0.05 is stricter; 0.10 is more permissive.
ROLLING_WINDOW = 200
ROLLING_STEP = 25
MAX_RESID_ADF_PVALUE = 0.10

# ----- Johansen settings -----
# JOHANSEN_DET_ORDER: deterministic term. 0 means constant term in the cointegration relation.
# JOHANSEN_K_AR_DIFF: lag differences used by the Johansen VAR setup.
JOHANSEN_DET_ORDER = 0
JOHANSEN_K_AR_DIFF = 1

raw = yf.download(
    tickers=[PRIMARY_SYMBOL, HEDGE_SYMBOL],
    interval=INTERVAL,
    period=PERIOD,
    auto_adjust=True,
    progress=False
)

close = raw["Close"][[PRIMARY_SYMBOL, HEDGE_SYMBOL]].copy()
close.columns = ["primary", "hedge"]
close = close.dropna()

df = close.copy()
df["log_primary"] = np.log(df["primary"])
df["log_hedge"] = np.log(df["hedge"])

print("===== DATA =====")
print("Primary:", PRIMARY_SYMBOL, PRIMARY_LABEL)
print("Hedge:  ", HEDGE_SYMBOL, HEDGE_LABEL)
print("Interval:", INTERVAL, "Period:", PERIOD)
print("Start:  ", df.index.min())
print("End:    ", df.index.max())
print("Rows:   ", len(df))

def safe_adf(series, regression="c"):
    series = pd.Series(series).dropna()
    if len(series) < 25:
        return {"stat": np.nan, "pvalue": np.nan, "lags": np.nan, "nobs": len(series)}
    try:
        stat, pvalue, lags, nobs, _, _ = adfuller(series, regression=regression, autolag="AIC")
        return {"stat": stat, "pvalue": pvalue, "lags": lags, "nobs": nobs}
    except Exception:
        return {"stat": np.nan, "pvalue": np.nan, "lags": np.nan, "nobs": len(series)}

def estimate_half_life(spread):
    spread = pd.Series(spread).dropna()
    lagged = spread.shift(1).dropna()
    delta = spread.diff().dropna()
    aligned = pd.concat([lagged, delta], axis=1).dropna()
    aligned.columns = ["lagged", "delta"]

    if len(aligned) < 25:
        return np.nan, np.nan

    model = sm.OLS(aligned["delta"], sm.add_constant(aligned["lagged"]))
    fit = model.fit()
    phi = fit.params["lagged"]

    if not np.isfinite(phi) or phi >= 0:
        return phi, np.nan

    half_life = -np.log(2.0) / phi
    return phi, half_life

primary_adf = safe_adf(df["log_primary"])
hedge_adf = safe_adf(df["log_hedge"])

x = sm.add_constant(df["log_hedge"])
ols = sm.OLS(df["log_primary"], x).fit()
alpha = ols.params["const"]
beta = ols.params["log_hedge"]
df["spread"] = df["log_primary"] - (alpha + beta * df["log_hedge"])
spread_adf = safe_adf(df["spread"])
phi, half_life = estimate_half_life(df["spread"])

eg_primary_on_hedge = coint(df["log_primary"], df["log_hedge"], trend="c", autolag="aic")
eg_hedge_on_primary = coint(df["log_hedge"], df["log_primary"], trend="c", autolag="aic")

johansen_input = df[["log_primary", "log_hedge"]].dropna().values
johansen = coint_johansen(johansen_input, det_order=JOHANSEN_DET_ORDER, k_ar_diff=JOHANSEN_K_AR_DIFF)

print("===== UNIT ROOT TESTS =====")
print(f"log primary ADF pvalue: {primary_adf['pvalue']:.4f}")
print(f"log hedge ADF pvalue:   {hedge_adf['pvalue']:.4f}")
print("Interpretation: for cointegration, individual price logs often should NOT be stationary; high p-values here are normal.")
print()
print("===== OLS SPREAD =====")
print(f"alpha:                 {alpha:.6f}  (intercept in the log-price hedge model)")
print(f"beta:                  {beta:.6f}  (hedge ratio in log-price space)")
print(f"spread ADF pvalue:     {spread_adf['pvalue']:.4f}  (lower means more evidence the residual spread is stationary)")
print(f"OU phi estimate:       {phi:.6f}  (negative means mean-reverting in this simple model)")
print(f"half-life (bars):      {half_life:.2f}  (rough bars for half a spread shock to decay)")
print()
print("===== ENGLE-GRANGER =====")
print(f"primary ~ hedge pvalue: {eg_primary_on_hedge[1]:.4f}")
print(f"hedge ~ primary pvalue: {eg_hedge_on_primary[1]:.4f}")
print("Interpretation: p < 0.05 is stronger evidence of cointegration; p < 0.10 is weaker but sometimes used for exploration.")
print()
print("===== JOHANSEN =====")
print("trace stats:     ", np.round(johansen.lr1, 4))
print("trace crit 95%:  ", np.round(johansen.cvt[:, 1], 4))
print("max eig stats:   ", np.round(johansen.lr2, 4))
print("max eig crit95%: ", np.round(johansen.cvm[:, 1], 4))
print("Interpretation: a Johansen statistic above its 95% critical value supports at least one cointegration relation.")

rolling_rows = []
for end_idx in range(ROLLING_WINDOW, len(df) + 1, ROLLING_STEP):
    window = df.iloc[end_idx - ROLLING_WINDOW:end_idx]
    x_win = sm.add_constant(window["log_hedge"])
    fit_win = sm.OLS(window["log_primary"], x_win).fit()
    alpha_win = fit_win.params["const"]
    beta_win = fit_win.params["log_hedge"]
    spread_win = window["log_primary"] - (alpha_win + beta_win * window["log_hedge"])
    adf_win = safe_adf(spread_win)
    eg_win = coint(window["log_primary"], window["log_hedge"], trend="c", autolag="aic")
    _, hl_win = estimate_half_life(spread_win)
    rolling_rows.append({
        "time": window.index[-1],
        "alpha": alpha_win,
        "beta": beta_win,
        "spread_adf_pvalue": adf_win["pvalue"],
        "engle_granger_pvalue": eg_win[1],
        "half_life_bars": hl_win
    })

rolling = pd.DataFrame(rolling_rows).set_index("time")
pass_ratio = (rolling["spread_adf_pvalue"] <= MAX_RESID_ADF_PVALUE).mean() if len(rolling) else np.nan
print()
print("===== ROLLING DIAGNOSTICS =====")
print(f"rolling windows:           {len(rolling)}")
print(f"ADF pass ratio <= {MAX_RESID_ADF_PVALUE:.2f}: {pass_ratio * 100:.2f}%")
print("Interpretation: higher pass ratio means the spread stays stationary more consistently through time.")
display(rolling.tail(10))
```

```python
# Visual diagnostics

plt.figure(figsize=(14, 6))
plt.plot(df.index, df["primary"], label=PRIMARY_LABEL)
plt.plot(df.index, df["hedge"], label=HEDGE_LABEL)
plt.title("Price series")
plt.xlabel("Time")
plt.ylabel("Price")
plt.grid(True)
plt.legend()
plt.show()

plt.figure(figsize=(14, 5))
plt.plot(df.index, df["spread"], label="OLS spread")
plt.axhline(df["spread"].mean(), color="black", linestyle="--", label="Mean")
plt.title("Log-price spread")
plt.xlabel("Time")
plt.ylabel("Spread")
plt.grid(True)
plt.legend()
plt.show()

zscore = (df["spread"] - df["spread"].rolling(ROLLING_WINDOW).mean()) / df["spread"].rolling(ROLLING_WINDOW).std()
plt.figure(figsize=(14, 5))
plt.plot(df.index, zscore, label="Rolling z-score")
plt.axhline(2.0, linestyle="--", color="tab:red", label="+2")
plt.axhline(-2.0, linestyle="--", color="tab:green", label="-2")
plt.axhline(0.0, linestyle="-", color="black", label="0")
plt.title("Spread z-score")
plt.xlabel("Time")
plt.ylabel("Z-score")
plt.grid(True)
plt.legend()
plt.show()

if len(rolling) > 0:
    fig, axes = plt.subplots(3, 1, figsize=(14, 12), sharex=True)

    axes[0].plot(rolling.index, rolling["beta"], label="Rolling beta")
    axes[0].set_title("Rolling hedge ratio")
    axes[0].set_ylabel("Beta")
    axes[0].grid(True)
    axes[0].legend()

    axes[1].plot(rolling.index, rolling["spread_adf_pvalue"], label="Rolling spread ADF pvalue")
    axes[1].axhline(MAX_RESID_ADF_PVALUE, linestyle="--", color="tab:red", label="ADF threshold")
    axes[1].set_title("Rolling residual stationarity")
    axes[1].set_ylabel("pvalue")
    axes[1].grid(True)
    axes[1].legend()

    axes[2].plot(rolling.index, rolling["engle_granger_pvalue"], label="Rolling Engle-Granger pvalue")
    axes[2].axhline(0.05, linestyle="--", color="tab:red", label="5% threshold")
    axes[2].set_title("Rolling Engle-Granger")
    axes[2].set_xlabel("Time")
    axes[2].set_ylabel("pvalue")
    axes[2].grid(True)
    axes[2].legend()

    plt.tight_layout()
    plt.show()
```
